## Example 1

```python
D=24
l=5
x = polygen(ZZ, 'x')
K.<d_K> = NumberField(x^2-D)
G=K.galois_group();
```

```python
getgam=getGamomega(D,l) ## give representatives of the class group and their stablizers in Gamma_0(l), if possible c shoud be small
print(getgam)
for i in range(0,len(getgam)):
    c1,d1=getgam[i][2][1]
    print('Computations will be done for c,d=',c1,d1, 'and omega=',getgam[i][1])
```

```python
p=7
M=100
S = Qp(p, prec = M, type = 'capped-rel', print_mode = 'series')
T.<d_p> = S.ext(x^2-D)
betap= T.primitive_root_of_unity()
```

```python
computing_units(getgam,l,M,p,T,betap) ## will print the minimal polynomial, and give the value of u_l(omega) and its ord_p, log_p ,log_beta
```
